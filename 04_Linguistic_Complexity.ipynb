### Surface complexity features:
- word count
- character count
- average word length



```python
def extract_surface_features(text):
    """
    Extract basic surface-level linguistic measurements.

    These are descriptive features and are not interpreted
    independently as evidence of psychological state.
    """

    text = str(text).strip()

    if not text:
        return {
            "word_count": 0,
            "char_count": 0,
            "avg_word_length": 0.0
        }

    words = text.split()

    word_count = len(words)
    char_count = len(text)

    avg_word_length = (
        np.mean([len(word) for word in words])
        if words else 0.0
    )

    return {
        "word_count": word_count,
        "char_count": char_count,
        "avg_word_length": avg_word_length
    }
```

### Dependency-based linguistic complexity
- dependency count
- dependency tree depth
- average dependency distance

#### What is dependency?
(eg) "The student solved the problem quickly."

A dependency parser tries to represent the grammatical relationships between words.

             solved
            /      \
        student    problem
          |           |
         The        the
                       \
                      quickly

student  → subject of → solved

problem  → object of  → solved

quickly  → modifier of → solved

```python
def extract_dependency_features(doc):
    """
    Extract syntactic complexity measures from a spaCy Doc.

    Features:
    - dependency_count:
        Number of syntactic dependency relations.
    - dependency_tree_depth:
        Maximum depth of the dependency structure.
    - avg_dependency_distance:
        Average linear distance between a token and its head.
    """

    tokens = [
        token for token in doc
        if not token.is_space and not token.is_punct
    ]

    if len(tokens) <= 1:
        return {
            "dependency_count": 0,
            "dependency_tree_depth": 0,
            "avg_dependency_distance": 0.0
        }

    # Number of dependency relations
    dependency_count = sum(
        1 for token in tokens
        if token.head != token
    )

    # --------------------------------------------------------
    # Dependency tree depth
    # --------------------------------------------------------

    def token_depth(token):
        depth = 0
        current = token

        visited = set()

        while current.head != current:
            if current.i in visited:
                break

            visited.add(current.i)
            depth += 1
            current = current.head

        return depth

    depths = [token_depth(token) for token in tokens]

    dependency_tree_depth = max(depths) if depths else 0

    # --------------------------------------------------------
    # Average dependency distance
    # --------------------------------------------------------

    distances = [
        abs(token.i - token.head.i)
        for token in tokens
        if token.head != token
    ]

    avg_dependency_distance = (
        float(np.mean(distances))
        if distances else 0.0
    )

    return {
        "dependency_count": dependency_count,
        "dependency_tree_depth": dependency_tree_depth,
        "avg_dependency_distance": avg_dependency_distance
    }
```

```python
for text in df["text"].head(5):

    doc = nlp(text)

    features = extract_dependency_features(doc)

    print("\nText:")
    print(text)

    print("Features:")
    print(features)
```

```python
texts = df["text"].astype(str).tolist()

surface_features = []
dependency_features = []

print("Processing", len(texts), "text segments...")

for i, (text, doc) in enumerate(
    zip(
        texts,
        nlp.pipe(texts, batch_size=64)
    )
):

    # Surface features
    surface = extract_surface_features(text)

    # Dependency features
    dependency = extract_dependency_features(doc)

    surface_features.append(surface)
    dependency_features.append(dependency)

    if (i + 1) % 2000 == 0:
        print(f"Processed {i + 1}/{len(texts)}")

print("Processing complete.")
```

```python
surface_df = pd.DataFrame(surface_features)
dependency_df = pd.DataFrame(dependency_features)

linguistic_features = pd.concat(
    [surface_df, dependency_df],
    axis=1
)

print("Feature table shape:", linguistic_features.shape)

print("\nFeatures:")
print(linguistic_features.columns.tolist())

print("\nFirst 5 rows:")
display(linguistic_features.head())
```

## Normalize structural measures

```python
complexity_features = linguistic_features.copy()

# Avoid division by zero
word_counts = complexity_features["word_count"].replace(0, np.nan)

complexity_features["dependency_density"] = (
    complexity_features["dependency_count"] / word_counts
)

complexity_features["dependency_density"] = (
    complexity_features["dependency_density"]
    .fillna(0.0)
)

complexity_features["dependency_distance_per_word"] = (
    complexity_features["avg_dependency_distance"]
    / word_counts
)

complexity_features["dependency_distance_per_word"] = (
    complexity_features["dependency_distance_per_word"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0.0)
)

print("Updated feature set:")
print(complexity_features.columns.tolist())

display(complexity_features.head())
```

```python
from scipy.stats import pearsonr
from sklearn.linear_model import LinearRegression

def partial_correlation(df, feature_x, feature_y, control_feature):
    data = df[[feature_x, feature_y, control_feature]].dropna()

    X_control = data[[control_feature]]

    # Residualize X
    model_x = LinearRegression()
    model_x.fit(X_control, data[feature_x])
    residual_x = data[feature_x] - model_x.predict(X_control)

    # Residualize Y
    model_y = LinearRegression()
    model_y.fit(X_control, data[feature_y])
    residual_y = data[feature_y] - model_y.predict(X_control)

    r, p = pearsonr(residual_x, residual_y)

    return r, p


for feature in features_to_control:
    r, p = partial_correlation(
        df_lc_reduced,
        feature,
        "avg_word_length",
        "word_count"
    )

    print(
        f"{feature} vs avg_word_length "
        f"(controlling word_count): "
        f"r = {r:.4f}, p = {p:.4e}"
    )
```

```python
from sklearn.linear_model import LinearRegression

syntactic_features = [
    "dependency_tree_depth",
    "avg_dependency_distance"
]

for feature in syntactic_features:

    data = df_lc_reduced[
        ["word_count", feature]
    ].dropna()

    X = data[["word_count"]]
    y = data[feature]

    model = LinearRegression()
    model.fit(X, y)

    predicted = model.predict(X)
    residual = y - predicted

    df_lc_reduced.loc[data.index, f"{feature}_length_residual"] = residual

    r2 = model.score(X, y)

    print(f"{feature}")
    print(f"R² explained by word_count = {r2:.4f}")
    print(f"Residual standard deviation = {residual.std():.4f}")
    print()
```

Four features were retained for the linguistic-complexity feature group: word count, average word length, dependency tree depth, and average dependency distance. Word count is retained primarily as an utterance-length/control variable. Dependency count and character count were excluded because they exhibited very high correlation with utterance length (r ≈ 0.99 and r ≈ 0.97, respectively), indicating substantial redundancy. Dependency density was also excluded because it is derived from dependency count and therefore provides little independent information. Dependency tree depth and average dependency distance were retained because, although both increase with utterance length, they retain substantial residual variation after accounting for length (45.5% and 71.8% of their variance, respectively, remains unexplained by word count).
